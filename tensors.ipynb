```python
import numpy as np
import torch as t

# list data to tensors5
list_data = [[1,2,3],[4,5,6],[7,8,9]]
x_data = t.tensor(list_data)

# np array to tensors
arr = np.array(list_data)
y_data = t.from_numpy(arr)

# retains the properties of x_data
x_ones = t.ones_like(x_data)
print(f'One Tensor: \n{x_ones}\n')

# overrides the datatype of x_data
x_rand = t.rand_like(x_data, dtype=float)
print(f'Random Tensor: \n{x_rand}\n')

# testing shape tuple
shape = (3,2)
rand_tensor = t.rand(shape)
ones_tensor = t.ones(shape)
zeros_tensor = t.zeros(shape)

print(f'Random Tensor: \n{rand_tensor}\n')
print(f'Ones Tensor: \n{ones_tensor}\n')
print(f'Zeros Tensor: \n{zeros_tensor}\n')

# testing Tensor attribute
tensor = t.rand(3,4)

print(f'Shape of tensor: {tensor.shape}')
print(f'Datatype of tensor: {tensor.dtype}')
print(f'Device tensor is stored on: {tensor.device}')

```

```python
import torch as t


if t.accelerator.is_available():
  tensor = tensor.to(t.accelerator.current_accelerator())
```

```python
import torch as t

# indexing (np style)

print(f"First row: {tensor[0]}")
print(f"First column: {tensor[:, 0]}")
print(f"Last column: {tensor[..., -1]}")
tensor[:,1] = 0
print(tensor)
```

```python
import torch as t


t1 = t.cat([tensor, tensor, tensor, tensor], dim=1)
print(t1)
```

```python
import torch as t


# This computes the matrix multiplication between two tensors. y1, y2, y3 will have the same value
# ``tensor.T`` returns the transpose of a tensor
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)

y3 = t.rand_like(y1)
t.matmul(tensor, tensor.T, out=y3)


# This computes the element-wise product. z1, z2, z3 will have the same value
z1 = tensor * tensor
z2 = tensor.mul(tensor)

z3 = t.rand_like(tensor)
t.mul(tensor, tensor, out=z3)

print(y1)
print(y2)
print(y3)
print(z1)
print(z2)
print(z3)
```
